```python
# Databricks notebook source

"""
NBA Game Data - Bronze Ingestion

Purpose:
    Ingest raw NBA game data from Kaggle dataset into a Delta Lake Bronze table.

Layer:
    Bronze

Source: 
    https://www.kaggle.com/datasets/nathanlauga/nba-games
"""
```

```python
BASE_URL = "https://api.balldontlie.io/v1"

GAMES_ENDPOINT = f"{BASE_URL}/games"
```

```python
import requests

response = requests.get(
    GAMES_ENDPOINT,
    params = {
        "per_page": 10
    },
    timeout = 30,
)

print(response.status_code)
print(response.text[:500])
```

```python
games_path = "/Volumes/workspace/default/sports_data/games.csv"

games_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(games_path)
)

display(games_df)
```

```python
print(f"Rows: {games_df.count()}")
print(f"Columns: {len(games_df.columns)}")
print("\nSchema:")
games_df.printSchema()
```

```python
display(games_df.limit(10))
```

```python
display(games_df.describe())
```

```python
games_df.printSchema()
```

```python
bronze_games = "bronze_nba_games"

(
    games_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(bronze_games)
)

display(spark.table(bronze_games))
```

```python
spark.sql("DESCRIBE DETAIL bronze_nba_games").show(truncate=False)
```
